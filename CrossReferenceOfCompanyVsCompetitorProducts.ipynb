```python
# Import dependencies
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import os
```

```python
# Display settings for better readability
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 10)  # Display only the first 10 rows in output
```

```python
# Reading all xlsx files
Cross_Reference_SWX_df = pd.read_excel('Competitve_Cross_Reference.xlsx')
Top60_Products_Review_Ratings_df = pd.read_excel('CombinedTop60_Products_by_Reviews_and_Ratings.xlsx')
Top50_Products_by_Cumul_Star_df = pd.read_excel('Top50_Competitor_products_by_CumulStarRating.xlsx')
```

```python
# Displaying column names of all dataframes
print('Columns of Cross_Reference_Company1_Competitor_df:', Cross_Reference_Company1_Competitor_df.columns)
print('Columns of Top60_Competitor_Products_Review_Ratings_df:', Top60_Competitor_Products_Review_Ratings_df.columns)
print('Columns of Top50_Competitor_Products_by_Cumul_Star_df:', Top50_Competitor_Products_by_Cumul_Star_df.columns)
```

```python
# Extracting the columns interested in
competitor_values = set(Cross_Reference_SWX_Competitor_df['Competitor'].dropna().unique())
product_model_number_top60 = set(Top60_Products_Review_Ratings_df['Product Model Number'].dropna().unique())
product_model_number_top50 = set(Top50_Products_by_Cumul_Star_df['Product Model Number'].dropna().unique())

# Find common values across all three sets
common_values = competitor_values & product_model_number_top60 & product_model_number_top50

# Print the common values
print(f"Common values across all three columns: {common_values}")
```

```python
# Extract unique values from each DataFrame's 'Product Model Number' column
unique_model_numbers_top60 = set(Top60_Products_Review_Ratings_df['Product Model Number'].dropna().unique())
unique_model_numbers_top50 = set(Top50_Products_by_Cumul_Star_df['Product Model Number'].dropna().unique())

# Combine the unique values from both DataFrames
combined_unique_model_numbers = unique_model_numbers_top60.union(unique_model_numbers_top50)

# Create a DataFrame from the combined unique values
merged_df = pd.DataFrame(combined_unique_model_numbers, columns=['Product Model Number'])

# Merge Product Titles from both DataFrames
merged_df = pd.merge(
    merged_df,
    Top60_Products_Review_Ratings_df[['Product Model Number', 'Product Title']],
    on='Product Model Number',
    how='left'
)

merged_df = pd.merge(
    merged_df,
    Top50_Products_by_Cumul_Star_df[['Product Model Number', 'Product Title']],
    on='Product Model Number',
    how='left',
    suffixes=('_Top60', '_Top50')
)

# Combine Product Titles into a single column, prioritizing non-null values
merged_df['Product Title'] = merged_df['Product Title_Top60'].combine_first(merged_df['Product Title_Top50'])

# Drop the temporary columns used for merging
merged_df.drop(columns=['Product Title_Top60', 'Product Title_Top50'], inplace=True)

# Sort the DataFrame by 'Product Model Number' in ascending order
merged_df.sort_values(by='Product Model Number', inplace=True)

# Create a new Excel file with the sorted and merged data
merged_df.to_excel('Product_Model_Numbers_Titles.xlsx', index=False)

print("New Excel file 'Product_Model_Numbers_Titles.xlsx' created with sorted merged unique product model numbers and their titles.")
```
